# 1. Model calls

For now we learn two ways to use the models in langchain. One is the **closed source models** and **their API**. The other is using **Ollama** models for **open source** **models**. 

```python
import os
from dotenv import load_dotenv
from IPython.display import display, Markdown
load_dotenv()

os.environ["OPENAI_API_KEY"]=os.getenv("OPENAI_API_KEY")
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
os.environ["GOOGLE_API_KEY"]=os.getenv("GOOGLE_API_KEY")
```

## Closed source models

### A. OpenAI

We have two ways to use **OPEN-AI**

```python
from langchain.chat_models import init_chat_model

model=init_chat_model("gpt-5")
response = model.invoke("Hello how are you?")
response
```

The other method is to use **ChatOpenAI**

```python
from langchain_openai import ChatOpenAI

model = ChatOpenAI(model="gpt-5")
response = model.invoke("Hello how are you?")
response
```

### B. Google Gemini

Again we have 2 ways to use Gemini

```python
from langchain.chat_models import init_chat_model

model = init_chat_model("google_genai:gemini-2.5-flash")
response = model.invoke("How did humen went to Mars?")
response.content
```

The other method

```python
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(model="google_genai:gemini-2.5-light")
response = model.invoke("How did humen went to Mars?")
response.content
```

### C. Groq

```python
from langchain.chat_models import init_chat_model

model = init_chat_model("groq:qwen/qwen3-32b")
response = model.invoke("Why do parrots talk?")
response
```

The other method is:

```python
from langchain_groq import ChatGroq

model = ChatGroq(model="google_genai:gemini-2.5-light")
response = model.invoke("How did humen went to Mars?")
response.content
```

## Open-source models

```python
from langchain_ollama import ChatOllama

model = ChatOllama(model="gemma3:4b", 
                   temperature=0)

response = model.invoke("How did humen went to mars?")

display(Markdown(response.content))
```

## Agentic calls

This type of call is usefull when we want to include the **tools** we want to use.

```python
from langchain.agents import create_agent
from langchain_ollama import ChatOllama

model = ChatOllama(model="gemma3:4b",
                   temperature=0)

agent = create_agent(model=model,
                     tools=[],
                     system_prompt="You are a helpful assistant")
```

```python
result = agent.invoke({"messages":[{"role":"user", "content":"what was the first voyage of humen to mars?"}]})
```

```python
display(Markdown(result["messages"][-1].content))
```
